```python
from transformers import AutoTokenizer, AutoModelForCausalLM
model_id = "Qwen/Qwen2.5-14B-Instruct-1M"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto")
```

```python
import token


s = [{"role": "user", "content": "hello my name is Mike. What is your name?"}]
s_enc = tokenizer.apply_chat_template(s, add_generation_prompt=True, tokenize=True, return_tensors='pt')
print(tokenizer.decode(s_enc[0]))
out = model.generate(s_enc)
```
